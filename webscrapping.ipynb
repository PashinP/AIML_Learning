```python
from bs4 import BeautifulSoup

page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)
    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div.quote")
    if not quotes:
        print("not a valid page now")
        break 
    with open(f"scraped_data/quotes{page_count}.html","w") as f:
        f.write(res.text)
        print(f"downaloaded data from {page_count}")
    page_count = page_count + 1
```

```python
import pandas as pd
page_count = 1
life_quotes = []
while page_count < 10:
    with open(f"scraped_data/quotes{page_count}.html","r") as f:
        html_content = f.read()
    soup = BeautifulSoup(html_content,"lxml")
    all_quotes = soup.select("div.quote")
    for q in all_quotes:
        all_tags = []
        for tag in  q.select(".tags .tag"):
            all_tags.append(tag.get_text())
        if "life" in all_tags:
            text =  q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text,author])
    page_count = page_count + 1
df = pd.DataFrame(life_quotes, columns=["Text", "Author"])
df.to_csv("cleaned_data/data.csv", index=False)
```
